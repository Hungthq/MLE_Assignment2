```python
def inspect_number_values(df):
    #inspect number values columns
    num_cols = df.select_dtypes(include=['object']).columns.tolist()
    for col in num_cols:
        if col in df.columns:
            print(f"Column: {col}")
            valid_number_pattern = r'^[+-]?(?:\d+\.?\d*|\.\d+)$'
            non_numeric_values = df[col][~df[col].str.contains(valid_number_pattern, regex=True, na=False)].unique()
            if len(non_numeric_values) > 0:
                print(f"Non-numeric values in column '{col}': {non_numeric_values}")
            #print(df[col][df[col].str.contains(r'^[+-]?(?:\d+\.?\d*|\.\d+)$', regex=True, na=False)].unique())
            print("\n")
```

```python
inspect_number_values(financial)
```

```python
# read features financial data
financial = pd.read_csv('data/features_financials.csv')
```

```python
#Clean up data in the financial dataset   
financial['Annual_Income'] = financial['Annual_Income'].str.replace('\..*', '', regex=True).astype(int)   
financial['Monthly_Inhand_Salary'] = (financial['Annual_Income'] / 12).round(2)
financial['Num_of_Loan'] = financial['Num_of_Loan'].str.replace('_', '', regex=True).astype(int).abs()

financial['Num_of_Delayed_Payment'] = financial['Num_of_Delayed_Payment'].str.replace('_', '', regex=True).astype(int).abs()
financial['Changed_Credit_Limit'] = financial['Changed_Credit_Limit'].str.replace('_', '0', regex=True).astype(float)
financial['Outstanding_Debt'] = financial['Outstanding_Debt'].str.replace('_', '', regex=True).astype(float)
financial['Amount_invested_monthly'] = financial['Amount_invested_monthly'].str.replace('_', '', regex=True).astype(float).round(2)
financial['Monthly_Balance'] = financial['Monthly_Balance'].str.replace('_', '', regex=True).round(2).astype(float)
financial['Credit_Utilization_Ratio'] = financial['Credit_Utilization_Ratio'].round(2)
financial['Total_EMI_per_month'] = financial['Total_EMI_per_month'].round(2)
financial['Monthly_Balance'] = financial['Monthly_Balance'].round(2)
```

```python
# convert '10 years 3 months' to months
def convert_credit_history_age_to_months(age_str):
    if pd.isnull(age_str):
        return np.nan
    years, months = 0, 0
    if 'Years' in age_str:
        years = int(age_str.split('Years')[0].strip())
    if 'Months' in age_str:
        months = int(age_str.split('Months')[0].split()[-1].strip())
    return years * 12 + months
```

```python
financial['Credit_History_Age_months'] = financial['Credit_History_Age'].apply(convert_credit_history_age_to_months)
```
